```python
file_list = ['data.xlsx', 'summary.docx', '1.csv']

def verify(file_list):
  list_2 = []
  for file in file_list:
    file = file.lower()
    if file in list_2:
      print(f'{file} Duplicate found') #mensagem de erro
      break #erro fatal
    list_2.append(file)
  else:
    print('All files are unique') #só executa se não tiver break

verify(file_list)
```
